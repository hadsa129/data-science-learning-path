```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
# Sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(10, 6)) # Set the figure size
plt.plot(x , y, label='sin(x)', color='blue', linewidth=2, linestyle='--') # plot the data,the attributes are :x(the x axis),y(the y axis),label(the label of the line),color(the color of the line),linewidth(the width of the line),linestyle(the style of the line)
plt.xlabel('X values', fontsize=12) # Set the x label
plt.ylabel('sin(X)', fontsize=12) # Set the y label
plt.title('Sine Function Plot', fontsize=14, fontweight='bold') # Set the title
plt.legend() # Add a legend :the legend is the label of the line
plt.grid(True, alpha=0.3) # Add a grid :the grid means the grid lines
plt.show() # Show the plot
```

```python
### Basic Plotting Workflow
fig, ax = plt.subplots(figsize=(10, 6))

# 1. Create data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# 2. Plot
ax.plot(x, y1, label='sin(x)', color='blue', linewidth=2, linestyle='-', marker='', alpha=0.8)
ax.plot(x, y2, label='cos(x)', color='red', linewidth=2, linestyle='--', marker='o', markersize=4, alpha=0.8)

# 3. Customize
ax.set_title('Sine and Cosine Waves', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('X values', fontsize=12)
ax.set_ylabel('Y values', fontsize=12)
ax.legend(loc='upper right', fontsize=10, frameon=True, shadow=True)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_xlim(0, 10)
ax.set_ylim(-1.5, 1.5)

# 4. Add annotations
ax.annotate('Maximum of sin(x)', xy=(np.pi/2, 1), xytext=(np.pi/2 + 1, 1.2),
            arrowprops=dict(arrowstyle='->', color='black', lw=1),
            fontsize=10, bbox=dict(boxstyle="round,pad=0.3", facecolor="yellow", alpha=0.5))

# 5. Save or show
plt.tight_layout()
plt.savefig('basic_plot.png', dpi=300, bbox_inches='tight')
plt.show()

print("\nFigure object attributes:")
print(f"Figure size: {fig.get_size_inches()}")
print(f"Number of axes: {len(fig.axes)}")
print(f"DPI: {fig.dpi}")
```

```python
import seaborn as sns
import pandas as pd

# Load sample dataset
tips = sns.load_dataset('tips')

# Create a boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(x='day', y='total_bill', data=tips, palette='Set2')
plt.title('Total Bill by Day', fontsize=14)
plt.xlabel('Day of Week', fontsize=12)
plt.ylabel('Total Bill ($)', fontsize=12)
plt.show()
```

```python
# Create sample datasets for visualization
np.random.seed(42)
n_samples = 1000

# Create a comprehensive dataset
data = {
    'Age': np.random.normal(40, 15, n_samples).clip(18, 100),
    'Income': np.random.lognormal(10, 0.5, n_samples),
    'Education_Years': np.random.choice([12, 14, 16, 18, 20], n_samples, p=[0.1, 0.3, 0.4, 0.15, 0.05]),
    'Job_Satisfaction': np.random.randint(1, 11, n_samples),
    'Department': np.random.choice(['Sales', 'Marketing', 'IT', 'HR', 'Finance'], n_samples),
    'Experience': np.random.exponential(10, n_samples).clip(0, 40),
    'Gender': np.random.choice(['Male', 'Female'], n_samples),
    'City': np.random.choice(['NYC', 'LA', 'Chicago', 'Houston', 'Phoenix'], n_samples),
    'Score_A': np.random.normal(70, 15, n_samples),
    'Score_B': 0.7 * np.random.normal(70, 15, n_samples) + 0.3 * np.random.normal(80, 10, n_samples),
    'Month': np.tile(np.arange(1, 13), n_samples//12 + 1)[:n_samples],
    'Sales': np.random.gamma(shape=2, scale=1000, size=n_samples) + np.sin(np.arange(n_samples)/10)*500
}

df = pd.DataFrame(data)

# Add some correlations
df['Income'] = df['Income'] + df['Education_Years'] * 5000 + np.random.normal(0, 10000, n_samples)
df['Job_Satisfaction'] = df['Job_Satisfaction'] + (df['Income'] > df['Income'].median()).astype(int)

print("Dataset created with shape:", df.shape)
print("\nFirst few rows:")
display(df.head())
print("\nData types:")
print(df.dtypes)
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# Basic histogram
sns.histplot(data=df, x='Age', ax=axes[0, 0], kde=True, color='skyblue')
axes[0, 0].set_title('Histogram with KDE')
axes[0, 0].axvline(df['Age'].mean(), color='red', linestyle='--', label=f'Mean: {df["Age"].mean():.1f}')
axes[0, 0].axvline(df['Age'].median(), color='green', linestyle='--', label=f'Median: {df["Age"].median():.1f}')
axes[0, 0].legend()
# Histogram with different bins :
sns.histplot(data=df, x='Income', bins=50, ax=axes[0, 1], color='salmon')
axes[0, 1].set_title('Histogram - 50 bins')
axes[0, 1].set_xlabel('Income ($)')

# Cumulative histogram
sns.histplot(data=df, x='Age', cumulative=True, stat='percent', 
             ax=axes[0, 2], color='lightgreen')
axes[0, 2].set_title('Cumulative Distribution')
axes[0, 2].set_ylabel('Cumulative Percent')
axes[0, 2].axhline(50, color='red', linestyle='--', alpha=0.5)

# Stacked histogram by category
sns.histplot(data=df, x='Income', hue='Gender', multiple='stack', 
             ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Stacked Histogram by Gender')

# Step histogram
sns.histplot(data=df, x='Education_Years', element='step', fill=False, 
             ax=axes[1, 1], color='purple')
axes[1, 1].set_title('Step Histogram')

# Density plot only
sns.kdeplot(data=df, x='Age', ax=axes[1, 2], color='orange', fill=True, alpha=0.5)
axes[1, 2].set_title('Kernel Density Estimate (KDE)')

plt.suptitle('Histogram Variations', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# What questions does a histogram answer?
questions = [
    "1. What is the most common value (mode)?",
    "2. Is the distribution symmetric or skewed?",
    "3. How spread out are the values?",
    "4. Are there multiple peaks (multimodal)?",
    "5. Are there gaps in the data?",
    "6. Are there outliers (isolated bars)?"
]

# Seaborn histplot with KDE
plt.figure(figsize=(12, 5))
sns.histplot(data=data, kde=True, bins=30, 
             stat='density', linewidth=2, 
             edgecolor='black', alpha=0.5)

# Add statistical annotations
mean = np.mean(data)
median = np.median(data)
std = np.std(data)

plt.axvline(mean, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean:.1f}')
plt.axvline(median, color='green', linestyle='--', linewidth=2, label=f'Median: {median:.1f}')
plt.axvline(mean + std, color='blue', linestyle=':', alpha=0.5, label=f'±1 SD')
plt.axvline(mean - std, color='blue', linestyle=':', alpha=0.5)

plt.legend()
plt.title('Histogram with Statistical Annotations')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
```

# boxplot

```python
print("=== BOX PLOT ===")
print("""
Purpose: Visualize distribution summary and detect outliers
# Use box plot when:
# 1. Comparing distributions across categories
# 2. Identifying outliers
# 3. Showing median and quartiles
# 4. Comparing multiple groups side-by-side
# 5. When data has outliers that would distort means

  
Data type: Numerical (with optional categorical grouping)
How to read:
  - Box: IQR (25th to 75th percentile)
  - Line in box: Median (50th percentile)
  - Whiskers: Typically 1.5 * IQR from quartiles
  - Points outside whiskers: Potential outliers
  
What questions it answers:
  - What is the central tendency?
  - How spread is the data?
  - Is the data symmetric or skewed?
  - Are there any outliers?
""")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Basic box plot
sns.boxplot(data=df, y='Income', ax=axes[0], color='lightblue')
axes[0].set_title('Income Distribution')
axes[0].set_ylabel('Income ($)')

# Calculate and display IQR statistics
Q1 = df['Income'].quantile(0.25)
Q3 = df['Income'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Add annotations
axes[0].axhline(Q1, color='red', linestyle='--', alpha=0.5, label=f'Q1: ${Q1:,.0f}')
axes[0].axhline(df['Income'].median(), color='green', linestyle='-', alpha=0.7, label=f'Median: ${df["Income"].median():,.0f}')
axes[0].axhline(Q3, color='blue', linestyle='--', alpha=0.5, label=f'Q3: ${Q3:,.0f}')
axes[0].axhline(upper_bound, color='orange', linestyle=':', alpha=0.5, label=f'Upper bound: ${upper_bound:,.0f}')
axes[0].legend()

# Grouped box plot
sns.boxplot(data=df, x='Department', y='Income', ax=axes[1], palette='Set3')
axes[1].set_title('Income by Department')
axes[1].set_xlabel('Department')
axes[1].set_ylabel('Income ($)')
axes[1].tick_params(axis='x', rotation=45)

# Notched box plot (shows confidence interval around median)
sns.boxplot(data=df, x='Gender', y='Age', notch=True, ax=axes[2], palette='coolwarm')
axes[2].set_title('Age by Gender (Notched Box Plot)')
axes[2].set_xlabel('Gender')
axes[2].set_ylabel('Age')

plt.tight_layout()
plt.show()

```

# Violin Plot

```python
print("=== VIOLIN PLOT ===")
print("""
Purpose: Combination of box plot and kernel density estimate
When to use:
# 1. You want to show full distribution shape
# 2. Comparing distributions across groups
# 3. Identifying multimodality
# 4. When box plots hide important distribution details
# 5. For smaller datasets where KDE is meaningful
  
Data type: Numerical (with categorical grouping)
How to read:
  - Shape: Shows density at different values
  - Width: More data at that value
  - Usually includes inner box plot
  
Advantages over box plot:
  - Shows distribution shape
  - Reveals multimodality
  - Better for comparing distributions
  
Disadvantages:
  - Can be misleading with small datasets
  - More complex to interpret
""")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Basic violin plot
sns.violinplot(data=df, y='Income', ax=axes[0], color='lightblue')
axes[0].set_title('Income Distribution (Violin Plot)')
axes[0].set_ylabel('Income ($)')

# Violin plot with split by category
sns.violinplot(data=df, x='Department', y='Income', ax=axes[1], palette='muted')
axes[1].set_title('Income by Department')
axes[1].set_xlabel('Department')
axes[1].set_ylabel('Income ($)')
axes[1].tick_params(axis='x', rotation=45)

# Violin plot with inner plot types
sns.violinplot(data=df, x='Gender', y='Age', inner='box', split=True, 
               ax=axes[2], palette='Set2')
axes[2].set_title('Age by Gender (with inner box plot)')
axes[2].set_xlabel('Gender')
axes[2].set_ylabel('Age')

plt.tight_layout()
plt.show()
```

# Scatter Plot

```python
print("=== SCATTER PLOT ===")
print("""
Purpose: Visualize relationship between two numerical variables
When to use:
# 1. Investigating correlation between two variables
# 2. Looking for patterns or trends
# 3. Identifying clusters or groups
# 4. Detecting outliers in 2D
# 5. Before performing regression analysis
  
Data type: Two numerical variables
How to read:
  - X-axis: First variable
  - Y-axis: Second variable
  - Points: Individual observations
  - Pattern: Positive/negative correlation, no correlation
  - Clusters: Groups of similar observations
  
Key attributes:
  - hue: Color by categorical variable
  - size: Size by numerical variable
  - style: Marker style by categorical variable
  - alpha: Transparency for overlapping points
""")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Basic scatter plot
axes[0, 0].scatter(df['Age'], df['Income'], alpha=0.6, s=20)
axes[0, 0].set_title('Basic Scatter: Age vs Income')
axes[0, 0].set_xlabel('Age')
axes[0, 0].set_ylabel('Income ($)')

# Add regression line
sns.regplot(data=df, x='Age', y='Income', ax=axes[0, 0], 
            scatter=False, color='red', line_kws={'linewidth': 2})

# Seaborn scatter with hue
sns.scatterplot(data=df, x='Age', y='Income', hue='Gender', 
                style='Gender', ax=axes[0, 1], s=50, alpha=0.7)
axes[0, 1].set_title('Scatter with Categorical Hue')
axes[0, 1].set_xlabel('Age')
axes[0, 1].set_ylabel('Income ($)')
axes[0, 1].legend(title='Gender', bbox_to_anchor=(1.05, 1))

# Scatter with size
sns.scatterplot(data=df, x='Experience', y='Job_Satisfaction', 
                size='Age', hue='Department', sizes=(20, 200),
                alpha=0.6, ax=axes[1, 0], palette='viridis')
axes[1, 0].set_title('Scatter with Size and Hue')
axes[1, 0].set_xlabel('Experience (Years)')
axes[1, 0].set_ylabel('Job Satisfaction')
axes[1, 0].legend(title='Department', bbox_to_anchor=(1.05, 1))

# Hexbin plot (for large datasets)
hb = axes[1, 1].hexbin(df['Age'], df['Income'], gridsize=30, cmap='Blues', bins='log')
axes[1, 1].set_title('Hexbin Plot (for Large Datasets)')
axes[1, 1].set_xlabel('Age')
axes[1, 1].set_ylabel('Income ($)')
plt.colorbar(hb, ax=axes[1, 1], label='Count')

plt.suptitle('Scatter Plot Variations', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

# LINE PLOT

```python
print("=== LINE PLOT ===")
print("""
Purpose: Visualize matrix data with colors
When to use:
# 1. Showing trends over time (time series)
# 2. Displaying continuous changes
# 3. Comparing multiple trends
# 4. Showing cumulative progress
# 5. Visualizing functions or mathematical relationships
  
Data type: 
 - X-axis: Ordered data (time, sequence, index)
 - Y-axis: Numerical values

What it shows
 - Trends over time or sequence
 - Changes in variable values
 - Patterns like seasonality, cycles
 - Comparisons between multiple series
""")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# Generate time series data
time = np.arange(100)
trend = 0.05 * time
seasonality = 10 * np.sin(2 * np.pi * time / 20)
noise = np.random.randn(100) * 2
data = trend + seasonality + noise

# 1. Basic line plot
axes[0, 0].plot(time, data, linewidth=1)
axes[0, 0].set_title('Basic Line Plot')
axes[0, 0].set_xlabel('Time')
axes[0, 0].set_ylabel('Value')

# 2. Line with markers
axes[0, 1].plot(time, data, marker='o', markersize=3, linewidth=1)
axes[0, 1].set_title('Line with Markers')

# 3. Multiple lines
axes[0, 2].plot(time, data, label='Original', linewidth=1)
axes[0, 2].plot(time, trend, label='Trend', linewidth=2, linestyle='--')
axes[0, 2].plot(time, seasonality + 50, label='Seasonality', linewidth=1.5, linestyle=':')
axes[0, 2].set_title('Multiple Lines')
axes[0, 2].legend()

# 4. Step plot
axes[1, 0].step(time, data, where='mid', linewidth=1)
axes[1, 0].set_title('Step Plot')

# 5. Area plot (filled)
axes[1, 1].fill_between(time, data, alpha=0.3)
axes[1, 1].plot(time, data, linewidth=1)
axes[1, 1].set_title('Area Plot')

# 6. Smooth line (moving average)
window = 10
smoothed = pd.Series(data).rolling(window=window).mean()
axes[1, 2].plot(time, data, alpha=0.3, label='Original')
axes[1, 2].plot(time[window-1:], smoothed[window-1:], linewidth=2, label=f'{window}-day MA')
axes[1, 2].set_title('With Moving Average')
axes[1, 2].legend()

plt.suptitle('Line Plot Variations', fontsize=16)
plt.tight_layout()
plt.show()
```

# Heatmap

```python
print("=== HEATMAP ===")
print("""
Purpose: Visualize matrix data with colors
When to use:
# 1. Displaying correlation matrices
# 2. Showing confusion matrices
# 3. Visualizing 2D data (like geographic data)
# 4. Identifying patterns in large matrices
# 5. Showing intensity (hot/cold spots)
  
Data type: Matrix/2D array of numerical values
How to read:
  - Colors represent magnitude
  - Typically red (high) to blue (low)
  - Look for patterns: clusters, diagonals, symmetry
  
Key attributes:
  - annot: Show values in cells
  - fmt: Format for annotations
  - cmap: Color map
  - center: Center value for diverging colormap
  - vmin/vmax: Value limits for colors
""")
```

```python
correlation_matrix = df.select_dtypes(include=[np.number]).corr()

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Basic heatmap
sns.heatmap(correlation_matrix, ax=axes[0], cmap='coolwarm', 
            center=0, annot=True, fmt='.2f', square=True,
            cbar_kws={"shrink": 0.8, "label": "Correlation"})
axes[0].set_title('Correlation Matrix Heatmap', fontsize=14)

# Enhanced heatmap with clustering
sns.heatmap(correlation_matrix, ax=axes[1], cmap='RdBu_r',
            center=0, annot=True, fmt='.2f', square=True,
            linewidths=0.5, linecolor='gray',
            cbar_kws={"shrink": 0.8, "label": "Correlation"})
axes[1].set_title('Correlation Matrix with Enhanced Styling', fontsize=14)

plt.tight_layout()
plt.show()
```

     ├──────────┼──────────┼──────────┤
Var3 │  -0.20   │   0.10   │   1.00   │  Dark blue = Strong negative
     └──────────┴──────────┴──────────┘
     
Interpretation Guidelines:
• Diagonal: Usually 1.0 (self-correlation)
• Symmetry: Correlation matrices are symmetric
• Color intensity: Magnitude of relationship
• Patterns: 
  - Clusters of similar colors → related variables
  - Checkerboard pattern → alternating relationships
  - Gradient patterns → ordered relationships

```python
# Exercise: Extract highly correlated pairs
print("=== EXERCISE: EXTRACT HIGHLY CORRELATED PAIRS ===")

# Create a mask for upper triangle
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))

# Extract pairs with high correlation
correlation_pairs = []
threshold = 0.7  # Define correlation threshold

for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        if not mask[i, j]:  # Skip diagonal and lower triangle
            corr_value = correlation_matrix.iloc[i, j]
            if abs(corr_value) > threshold:
                var1 = correlation_matrix.columns[i]
                var2 = correlation_matrix.columns[j]
                correlation_pairs.append({
                    'Variable 1': var1,
                    'Variable 2': var2,
                    'Correlation': corr_value,
                    'Strength': 'Strong Positive' if corr_value > 0.7 else 'Strong Negative'
                })

# Create DataFrame of high correlations
if correlation_pairs:
    high_corr_df = pd.DataFrame(correlation_pairs).sort_values('Correlation', ascending=False)
    print(f"Found {len(high_corr_df)} pairs with |correlation| > {threshold}:")
    display(high_corr_df)
    
    # Visualize just the high correlations
    plt.figure(figsize=(8, 6))
    high_corr_pivot = high_corr_df.pivot(index='Variable 1', columns='Variable 2', values='Correlation')
    sns.heatmap(high_corr_pivot, annot=True, cmap='coolwarm', center=0, 
                fmt='.2f', cbar_kws={"shrink": 0.8, "label": "Correlation"})
    plt.title(f'High Correlation Pairs (|r| > {threshold})', fontsize=14)
    plt.tight_layout()
    plt.show()
else:
    print(f"No pairs found with correlation > {threshold}")
```

# Joint Plot

```python
print("=== JOINT PLOT ===")
print("""
Purpose: Detailed visualization of relationship between two variables
When to use:
  - Focus on relationship between two key variables
  - Show both scatter and marginal distributions
  - Add regression or density estimates
  
Data type: Two numerical variables
How to read:
  - Main plot: Scatter/hexbin/kde of two variables
  - Top: Distribution of x variable
  - Right: Distribution of y variable
  - May include regression line
""")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Basic joint plot
g1 = sns.jointplot(data=df, x='Age', y='Income', kind='scatter', 
                   alpha=0.6, s=30, ax=axes[0, 0])
axes[0, 0].set_title('Scatter Joint Plot')
# Note: jointplot creates its own figure, so we need to handle it differently

# Clear axes and create proper layout
fig.clear()
plt.close()
# Create separate joint plots
print("Creating Joint Plots...")

# 1. Scatter with regression
g1 = sns.jointplot(data=df, x='Age', y='Income', kind='reg',
                   scatter_kws={'alpha': 0.5, 's': 20},
                   line_kws={'color': 'red', 'linewidth': 2},
                   marginal_kws={'bins': 30, 'kde': True})
g1.fig.suptitle('Joint Plot with Regression', y=1.02, fontsize=14)
g1.ax_joint.set_xlabel('Age')
g1.ax_joint.set_ylabel('Income ($)')

# 2. Hexbin joint plot
g2 = sns.jointplot(data=df, x='Age', y='Income', kind='hex',
                   cmap='Blues', gridsize=30)
g2.fig.suptitle('Hexbin Joint Plot', y=1.02, fontsize=14)
g2.ax_joint.set_xlabel('Age')
g2.ax_joint.set_ylabel('Income ($)')

# 3. KDE joint plot
g3 = sns.jointplot(data=df, x='Age', y='Income', kind='kde',
                   cmap='Reds', fill=True, levels=15)
g3.fig.suptitle('KDE Joint Plot', y=1.02, fontsize=14)
g3.ax_joint.set_xlabel('Age')
g3.ax_joint.set_ylabel('Income ($)')

# 4. Scatter with hue
# For this we need to use JointGrid
g4 = sns.JointGrid(data=df.sample(300), x='Age', y='Income', hue='Gender')
g4.plot_joint(sns.scatterplot, s=30, alpha=0.7)
g4.plot_marginals(sns.kdeplot, fill=True, alpha=0.5)
g4.fig.suptitle('Joint Plot with Gender Hue', y=1.02, fontsize=14)
g4.ax_joint.set_xlabel('Age')
g4.ax_joint.set_ylabel('Income ($)')
g4.ax_joint.legend(title='Gender')

plt.tight_layout()
plt.show()
```

# PIE CHART

```python
# Use pie chart when:
# 1. Showing parts of a whole (percentages)
# 2. Few categories (2-6 ideally)
# 3. Simple comparison of proportions
# 4. When relative sizes are important
# 5. For high-level summary

# Example: Market share
companies = ['Apple', 'Samsung', 'Google', 'Other']
market_share = [45, 30, 15, 10]

fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(market_share, labels=companies, autopct='%1.1f%%')
plt.show()
```

```python
# Create data
categories = ['Rent', 'Food', 'Transport', 'Entertainment', 'Savings', 'Other']
expenses = [1200, 800, 400, 300, 500, 200]

fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# 1. Basic pie chart
axes[0, 0].pie(expenses, labels=categories, autopct='%1.1f%%')
axes[0, 0].set_title('Basic Pie Chart')

# 2. Exploded pie chart
explode = [0.1, 0, 0, 0, 0, 0]  # "Explode" the first slice
axes[0, 1].pie(expenses, labels=categories, autopct='%1.1f%%', 
               explode=explode, shadow=True)
axes[0, 1].set_title('Exploded Pie Chart')

# 3. Donut chart
wedges, texts, autotexts = axes[0, 2].pie(expenses, labels=categories, 
                                          autopct='%1.1f%%')
# Add a circle at the center
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
axes[0, 2].add_artist(centre_circle)
axes[0, 2].set_title('Donut Chart')

# 4. Start angle (rotate)
axes[1, 0].pie(expenses, labels=categories, autopct='%1.1f%%',
               startangle=90)
axes[1, 0].set_title('Rotated (startangle=90)')

# 5. Custom colors
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#FFD700', '#C9A0DC']
axes[1, 1].pie(expenses, labels=categories, autopct='%1.1f%%',
               colors=colors)
axes[1, 1].set_title('Custom Colors')

# 6. Percentage outside
wedges, texts = axes[1, 2].pie(expenses, labels=categories)
axes[1, 2].set_title('Percentages Outside')

# Add percentages outside the pie
for i, (wedge, expense) in enumerate(zip(wedges, expenses)):
    ang = (wedge.theta2 - wedge.theta1)/2. + wedge.theta1
    y = np.sin(np.deg2rad(ang))
    x = np.cos(np.deg2rad(ang))
    
    percentage = expense/sum(expenses)*100
    horizontalalignment = {-1: "right", 1: "left"}[int(np.sign(x))]
    connectionstyle = f"angle,angleA=0,angleB={ang}"
    
    axes[1, 2].annotate(f'{percentage:.1f}%', xy=(x, y), xytext=(1.3*np.sign(x), 1.3*y),
                       horizontalalignment=horizontalalignment,
                       arrowprops=dict(arrowstyle="-", connectionstyle=connectionstyle))

plt.suptitle('Pie Chart Variations', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
print("""
PIE CHART PITFALLS - WHEN TO AVOID:

1. Too many slices (>6)
   Problem: Hard to compare small slices
   Solution: Group small categories as "Other"

2. Similar sized slices
   Problem: Hard to see differences
   Solution: Use bar chart instead

3. Need precise comparisons
   Problem: Humans bad at comparing angles
   Solution: Use bar chart with values

4. Multiple pies for comparison
   Problem: Hard to compare across pies
   Solution: Use stacked bar chart

5. Showing changes over time
   Problem: Multiple pies confusing
   Solution: Use line or area chart
""")

# Bad examples
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

# 1. Too many categories
too_many = [f'Item {i}' for i in range(15)]
values = np.random.randint(1, 20, 15)
ax1.pie(values, labels=too_many)
ax1.set_title('Too Many Categories (Unreadable)')

# 2. Similar sized slices
similar = [23, 24, 25, 28]
labels = ['A', 'B', 'C', 'D']
ax2.pie(similar, labels=labels, autopct='%1.1f%%')
ax2.set_title('Similar Sizes (Hard to Compare)')

# 3. Multiple pies (bad for comparison)
data1 = [30, 40, 30]
data2 = [25, 50, 25]
labels = ['X', 'Y', 'Z']

# Instead of two pies, use grouped bars
categories = ['X', 'Y', 'Z']
x = np.arange(len(categories))
width = 0.35

ax3.bar(x - width/2, data1, width, label='Group 1')
ax3.bar(x + width/2, data2, width, label='Group 2')
ax3.set_xticks(x)
ax3.set_xticklabels(categories)
ax3.set_ylabel('Percentage')
ax3.set_title('Better: Grouped Bar Chart')
ax3.legend()

plt.tight_layout()
plt.show()
```

# Time Series Visualizations

```python
print("=== TIME SERIES PLOTS ===")
print("""
Purpose: Visualize data over time
When to use:
  - Track changes over time
  - Identify trends, seasonality, cycles
  - Compare multiple time series
  
Data type: Time on x-axis, numerical on y-axis
Types:
  - Line plot: Basic time series
  - Area plot: Emphasize volume
  - Seasonal plot: Compare same periods
  - Rolling average: Smooth trends
  
How to read:
  - Trend: Overall direction
  - Seasonality: Regular patterns
  - Cycles: Longer-term fluctuations
  - Noise: Random variation
""")
```

```python
# Create time series data
np.random.seed(42)
dates = pd.date_range('2023-01-01', periods=365, freq='D')
time_series = pd.DataFrame({
    'Date': dates,
    'Temperature': 20 + 10 * np.sin(2 * np.pi * np.arange(365)/365) + np.random.normal(0, 3, 365),
    'Sales': 1000 + 500 * np.sin(2 * np.pi * np.arange(365)/365 * 12) + np.random.normal(0, 200, 365),
    'Visitors': 500 + 200 * np.cos(2 * np.pi * np.arange(365)/365 * 26) + np.random.normal(0, 50, 365)
})
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Basic line plot
axes[0, 0].plot(time_series['Date'], time_series['Temperature'], linewidth=1, alpha=0.8)
axes[0, 0].set_title('Daily Temperature Over Year', fontsize=12)
axes[0, 0].set_xlabel('Date')
axes[0, 0].set_ylabel('Temperature (°C)')
axes[0, 0].tick_params(axis='x', rotation=45)

# Multiple time series
axes[0, 1].plot(time_series['Date'], time_series['Sales'], label='Sales', alpha=0.8)
axes[0, 1].plot(time_series['Date'], time_series['Visitors'], label='Visitors', alpha=0.8)
axes[0, 1].set_title('Sales and Visitors Over Time', fontsize=12)
axes[0, 1].set_xlabel('Date')
axes[0, 1].set_ylabel('Count')
axes[0, 1].legend()
axes[0, 1].tick_params(axis='x', rotation=45)

# Area plot
axes[1, 0].fill_between(time_series['Date'], 
                        time_series['Sales'] - time_series['Sales'].std(),
                        time_series['Sales'] + time_series['Sales'].std(),
                        alpha=0.3, color='skyblue', label='±1 SD')
axes[1, 0].plot(time_series['Date'], time_series['Sales'], linewidth=2, color='blue', label='Sales')
axes[1, 0].set_title('Sales with Confidence Band', fontsize=12)
axes[1, 0].set_xlabel('Date')
axes[1, 0].set_ylabel('Sales')
axes[1, 0].legend()
axes[1, 0].tick_params(axis='x', rotation=45)

# Rolling average
rolling_window = 30
time_series['Sales_MA'] = time_series['Sales'].rolling(window=rolling_window).mean()
axes[1, 1].plot(time_series['Date'], time_series['Sales'], alpha=0.3, label='Daily Sales')
axes[1, 1].plot(time_series['Date'], time_series['Sales_MA'], linewidth=2, color='red', 
                label=f'{rolling_window}-day Moving Average')
axes[1, 1].set_title('Sales with Moving Average', fontsize=12)
axes[1, 1].set_xlabel('Date')
axes[1, 1].set_ylabel('Sales')
axes[1, 1].legend()
axes[1, 1].tick_params(axis='x', rotation=45)

plt.suptitle('Time Series Visualizations', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
