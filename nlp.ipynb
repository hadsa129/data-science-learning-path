## Text Normalization

Transforming text into a canonical form to reduce variability while preserving meaning.

Common Techniques:
1. Lowercasing: Convert all text to lowercase
2. Stemming: Reduce words to their root form (aggressive)
3. Lemmatization: Reduce words to dictionary form (linguistically accurate)
4. Removing Noise: URLs, HTML tags, special characters
5. Expanding Contractions: ”don’t” → ”do not”
6. Correcting Spelling: Fix common spelling errors

What is NLTK?
Natural Language Toolkit (nltk) is a Python package for NLP


**Stemming**
Stemming removes affixes to reduce to tokens to their base form (stems)

For example:

automates, automating and automatic could be stemmed to automat


There are 3 types of commonly used stemmers, and each consists of slightly different rules for systematically replacing affixes in tokens. In general, Lancaster stemmer stems the most aggresively, i.e. removing the most suffix from the tokens, followed by Snowball and Porter

Porter Stemmer:

Most commonly used stemmer and the most gentle stemmers
The most computationally intensive of the algorithms (Though not by a very significant margin)
The oldest stemming algorithm in existence
Snowball Stemmer:

Universally regarded as an improvement over the Porter Stemmer
Slightly faster computation time than the Porter Stemmer
Lancaster Stemmer:

Very aggressive stemming algorithm
With Porter and Snowball Stemmers, the stemmed representations are usually fairly intuitive to a reader
With Lancaster Stemmer, shorter tokens that are stemmed will become totally obfuscated
The fastest algorithm and will reduce the vocabulary
However, if one desires more distinction between tokens, Lancaster Stemmer is not recommended

**Algorithms for stemming**
- Manual Stemming (Not used)
- Automated Stemming
    - Lookup table
    - Rule based stemmers (affix)
        - Suffix
        - Prefix
    - Stochastic algorithms
    - Hybrid approaches

**Lemmatization**
Lemmatization aims to remove inflectional endings only and to return the base or dictionary form of a word, which is known as the lemma.

This is doing things properly with the use of a vocabulary and morphological analysis of words

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
import re
from autocorrect import Speller

# Download required NLTK data
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt')

# Sample text
text = "The cats are running quickly. Don't you think they're amazing?"

# 1. Lowercasing: lowercase all text
lower_text = text.lower()
print("Lowercasing:", lower_text)

# 2. Stemming: reduce words to their root form
stemmer = PorterStemmer()
tokens = word_tokenize(text)
stems = [stemmer.stem(token) for token in tokens]
print("\nStemming:", ' '.join(stems))

# 3. Lemmatization: reduce words to their dictionary form
lemmatizer = WordNetLemmatizer()
# First lemmatize as verbs
lemmatized = [lemmatizer.lemmatize(token, pos='v') for token in tokens]
# Then lemmatize as nouns (some words might be better lemmatized as nouns)
lemmatized = [lemmatizer.lemmatize(token, pos='n') for token in lemmatized]
print("\nLemmatization:", ' '.join(lemmatized))

# 4. Remove special characters and numbers
clean_text = re.sub(r'[^a-zA-Z\s]', ' ', text)
print("\nRemoved special chars:", clean_text)

# 5. Expand contractions (simplified)
contractions = {
    "don't": "do not",
    "doesn't": "does not",
    "isn't": "is not",
    "they're": "they are"
}

expanded_text = text
for contraction, expansion in contractions.items():
    expanded_text = expanded_text.replace(contraction, expansion)
print("\nExpanded contractions:", expanded_text)

# 6. Spell correction
try:
    spell = Speller(lang='en')
    misspelled = "I luv natural langage procesing"
    corrected = spell(misspelled)
    print("\nSpell correction:", corrected)
except Exception as e:
    print(f"\nSpell correction error (make sure autocorrect is installed): {e}")
```

## Stop Words Removal

Stop Words: Common words that carry little semantic meaning (the, is, at, which, etc.)

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# Download required NLTK data
nltk.download('stopwords')
nltk.download('punkt')

# Sample text
text = "The quick brown fox jumps over the lazy dog"
tokens = word_tokenize(text.lower())

# NLTK stop words :nltk delete stop words based on english language
nltk_stop_words = set(stopwords.words('english'))
filtered_nltk = [word for word in tokens if word not in nltk_stop_words]
print("NLTK stop words removed:", ' '.join(filtered_nltk))

# Scikit-learn stop words 
filtered_sklearn = [word for word in tokens if word not in ENGLISH_STOP_WORDS]
print("Scikit-learn stop words removed:", ' '.join(filtered_sklearn))

# Custom stop words
custom_stop_words = {'quick', 'brown', 'lazy'}
filtered_custom = [word for word in tokens if word not in custom_stop_words]
print("Custom stop words removed:", ' '.join(filtered_custom))

# When NOT to remove stop words:
# - Question Answering: "to be or not to be" - all words important
# - Sentiment Analysis: "not good" - "not" is crucial
# - Phrase detection: "New York" - both words needed
# - Machine Translation: Stop words are essential for grammar
# - Text Generation: Stop words maintain fluency
```

## Part-of-Speech Tagging

Assigning grammatical categories (noun, verb,adjective, etc.) to each word in a sentence.



```python
import nltk
import spacy
from spacy import displacy

# Download required NLTK data
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')

# Load spaCy model :this model is used to extract information from the text
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Please install the spaCy English model first:")
    print("python -m spacy download en_core_web_sm")
    nlp = None

sentences = [
    "The quick brown fox jumps over the lazy dog.",
    "I love natural language processing!",
    "She is reading a book about machine learning."
]

# NLTK POS Tagging
print("NLTK POS Tagging:")
for sentence in sentences:
    tokens = nltk.word_tokenize(sentence)
    tags = nltk.pos_tag(tokens)
    print(f"\n{sentence}")
    for token, tag in tags:
        print(f"{token:15} -> {tag}")

if nlp:
    print("\n" + "=" * 60)
    print("spaCy POS Tagging (with dependencies):")
    for sentence in sentences:
        doc = nlp(sentence)
        print(f"\n{sentence}")
        for token in doc:
            print(f"{token.text:15} -> {token.pos_:10} {token.dep_:15} {token.head.text}")

    # Visualize dependencies
    print("\nDependency Visualization:")
    doc = nlp("The quick brown fox jumps over the lazy dog.")
    displacy.render(doc, style="dep", jupyter=True)

    # Using POS tags for specific tasks
    def extract_noun_phrases(sentence):
        """Extract noun phrases using POS patterns."""
        doc = nlp(sentence)
        noun_phrases = []
        current_phrase = []
        
        for token in doc:
            if token.pos_ in ['NOUN', 'PROPN', 'ADJ', 'DET']:
                current_phrase.append(token.text)
            elif current_phrase:
                noun_phrases.append(' '.join(current_phrase))
                current_phrase = []
        
        if current_phrase:
            noun_phrases.append(' '.join(current_phrase))
            
        return noun_phrases

    print("\nNoun phrases extraction:")
    for sentence in sentences:
        phrases = extract_noun_phrases(sentence)
        print(f"{sentence} -> {phrases}")
else:
    print("\nSpaCy model not loaded. Skipping spaCy-related functionality.")
```

## Bag of Words (BoW)

Represent text as a vector of word counts, ignoring word order and grammar.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

documents = [
    "I love natural language processing",
    "Natural language processing is amazing",
    "I love machine learning and natural language processing",
    "Machine learning is a subset of artificial intelligence"
]

# 1. Basic Bag of Words
print("="*50)
print("1. Basic Bag of Words")
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
df_bow = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)
print(df_bow)

# 2. N-grams (capturing some word order)
print("\n" + "="*50)
print("2. Bag of Words with Bigrams (1-2 grams)")
bigram_vectorizer = CountVectorizer(ngram_range=(1, 2))
X_bigram = bigram_vectorizer.fit_transform(documents)
df_bigram = pd.DataFrame(
    X_bigram.toarray(),
    columns=bigram_vectorizer.get_feature_names_out()
)
# Print selected columns for better readability
print(df_bigram[['love', 'natural language', 'language processing', 'machine learning']])

# 3. TF-IDF representation
print("\n" + "="*50)
print("3. TF-IDF Representation")
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(documents)
df_tfidf = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)
print(df_tfidf.round(2))  # Round to 2 decimal places for better readability
```

## Term Frequency-Inverse Document Frequency (TF-IDF)

A numerical statistic that reflects how important a word is to a document in a collection.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def manual_tf_idf(corpus):
    documents = [doc.split() for doc in corpus]
    vocab = sorted(set(word for doc in documents for word in doc))
    
    # Calculate TF (Term Frequency)
    tf = np.zeros((len(documents), len(vocab)))
    for i, doc in enumerate(documents):
        for word in doc:
            tf[i, vocab.index(word)] += 1
        tf[i] = tf[i] / len(doc)  # Normalize
    
    # Calculate IDF (Inverse Document Frequency)
    doc_freq = np.zeros(len(vocab))
    for j, word in enumerate(vocab):
        for doc in documents:
            if word in doc:
                doc_freq[j] += 1
    idf = np.log(len(documents) / (doc_freq + 1e-6))
    
    # Calculate TF-IDF
    tf_idf = tf * idf
    return tf_idf, vocab

# Sample corpus
corpus = [
    "the cat sat on the mat",
    "the dog sat on the log",
    "cats and dogs are pets",
    "pets are amazing companions"
]

# Manual TF-IDF calculation
print("="*50)
print("1. Manual TF-IDF Calculation")
tf_idf_manual, vocab = manual_tf_idf(corpus)
for i, doc in enumerate(corpus):
    print(f"\nDocument {i+1}: {doc}")
    for j, word in enumerate(vocab):
        if tf_idf_manual[i, j] > 0:
            print(f"  {word}: {tf_idf_manual[i, j]:.3f}")

# Using scikit-learn
print("\n" + "="*50)
print("2. Scikit-learn TF-IDF")
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)
feature_names = vectorizer.get_feature_names_out()

for i, doc in enumerate(corpus):
    print(f"\nDocument {i+1}: {doc}")
    feature_index = X[i, :].nonzero()[1]
    for j in feature_index:
        print(f"  {feature_names[j]}: {X[i, j]:.3f}")

# Document similarity
print("\n" + "="*50)
print("3. Document Similarity Matrix")
similarity_matrix = cosine_similarity(X)
print("\nCosine Similarity Matrix:")
print(np.round(similarity_matrix, 2))

# Interpretation
print("\n" + "="*50)
print("4. Insights from TF-IDF:")
print("1. Common words (the, on, are) get low IDF scores")
print("2. Unique words (cats, dogs, amazing, companions) get high IDF scores")
print("3. Documents with similar word distributions have high cosine similarity")
print("4. Documents 1 and 2 are similar (both about animals sitting on things)")
print("5. Documents 3 and 4 are similar (both about pets)")
print("6. The diagonal shows self-similarity (always 1.0)")
```

## Word Embeddings

Instead of sparse vectors, learn dense vectors that capture semantic meaning.

### Word2Vec
Word2Vec is a popular method for learning word embeddings.
A neural network model that learns word embeddings by predicting words from their context (CBOW) or context from words (Skip-gram)


```python
import nltk
from nltk.corpus import brown
import numpy as np
import gensim
from gensim.models import Word2Vec
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Download required NLTK data
nltk.download('brown')

# Load sample corpus
sentences = brown.sents()[:1000]  # First 1000 sentences
print(f"Number of sentences: {len(sentences)}")
print(f"Sample sentence: {sentences[0]}")

# Train Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,      # Dimension of word vectors
    window=5,            # Context window size
    min_count=2,         # Ignore words with freq < 2
    workers=4,           # Number of CPU cores
    sg=1                 # 1 for skip-gram, 0 for CBOW
)

print(f"\nVocabulary size: {len(model.wv)}")

# Get word vector
vector_cat = model.wv['cat']
print(f"\nVector for 'cat' (first 10 dimensions): {vector_cat[:10]}")
print(f"Vector shape: {vector_cat.shape}")

# Find similar words
similar_words = model.wv.most_similar('man', topn=5)
print("\nWords similar to 'man':")
for word, score in similar_words:
    print(f"  {word}: {score:.3f}")

# Word analogy: king - man + woman = ?
result = model.wv.most_similar(
    positive=['woman', 'king'],
    negative=['man'],
    topn=3
)
print("\nking - man + woman = ?")
for word, score in result:
    print(f"  {word}: {score:.3f}")

# Visualize embeddings (2D projection)
words = ['king', 'queen', 'man', 'woman', 'prince', 'princess', 
         'car', 'truck', 'vehicle', 'bicycle', 'happy', 'sad', 'angry']

# Get vectors
word_vectors = np.array([model.wv[word] for word in words])

# Reduce to 2D for visualization
pca = PCA(n_components=2)
vectors_2d = pca.fit_transform(word_vectors)

# Plot
plt.figure(figsize=(12, 10))
plt.scatter(vectors_2d[:, 0], vectors_2d[:, 1])

# Add labels
for i, word in enumerate(words):
    plt.annotate(
        word,
        xy=(vectors_2d[i, 0], vectors_2d[i, 1]),
        xytext=(5, 2),
        textcoords='offset points'
    )

plt.title('Word Embeddings Visualization (PCA)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Save and load model
model.save("word2vec.model")
loaded_model = Word2Vec.load("word2vec.model")

# Using pre-trained embeddings (commented out as it requires downloading)
# print("\nLoading pre-trained Google News vectors (takes a minute)...")
# import gensim.downloader as api
# wv = api.load('word2vec-google-news-300')
# print("Similar to 'computer':", wv.most_similar('computer', topn=3))
```

### GloVe (Global Vectors for Word Representation)

Creates word embeddings by factorizing the word co-occurrence matrix, capturing global statistical information.

```python
import numpy as np
from gensim.models import KeyedVectors
from sklearn.decomposition import TruncatedSVD
from numpy.linalg import norm

def create_cooccurrence_matrix(corpus, window_size=2):
    """Create word co-occurrence matrix."""
    # Build vocabulary
    vocab = set()
    for sentence in corpus:
        vocab.update(sentence)
    vocab = sorted(vocab)
    word2idx = {word: i for i, word in enumerate(vocab)}
    
    # Initialize co-occurrence matrix
    n = len(vocab)
    cooccurrence = np.zeros((n, n))
    
    # Count co-occurrences
    for sentence in corpus:
        for i, word in enumerate(sentence):
            center_idx = word2idx[word]
            # Define context window
            start = max(0, i - window_size)
            end = min(len(sentence), i + window_size + 1)
            
            for j in range(start, end):
                if i != j:  # Don't count the word itself
                    context_idx = word2idx[sentence[j]]
                    cooccurrence[center_idx, context_idx] += 1
    
    return cooccurrence, vocab, word2idx

# Example corpus
corpus = [
    ["the", "cat", "sat", "on", "the", "mat"],
    ["the", "dog", "played", "with", "the", "ball"],
    ["cats", "and", "dogs", "are", "pets"],
    ["pets", "are", "good", "companions"]
]

# Create co-occurrence matrix
cooccurrence, vocab, word2idx = create_cooccurrence_matrix(corpus)
print("Vocabulary:", vocab)
print("\nCo-occurrence matrix (first 5x5):")
print(cooccurrence[:5, :5].astype(int))

# Simple GloVe-like embedding using matrix factorization
log_cooccurrence = np.log(cooccurrence + 1)  # Add 1 to avoid log(0)
svd = TruncatedSVD(n_components=10)  # 10-dimensional embeddings
embeddings = svd.fit_transform(log_cooccurrence)

print(f"\nWord embeddings shape: {embeddings.shape}")
print("\nEmbeddings for 'cat' and 'dog':")
print(f"  cat: {embeddings[word2idx['cat']][:5]}...")
print(f"  dog: {embeddings[word2idx['dog']][:5]}...")

# Similarity calculation
def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (norm(vec1) * norm(vec2))

cat_vec = embeddings[word2idx['cat']]
dog_vec = embeddings[word2idx['dog']]
the_vec = embeddings[word2idx['the']]

print("\nCosine similarities:")
print(f"  cat-dog: {cosine_similarity(cat_vec, dog_vec):.3f}")
print(f"  cat-the: {cosine_similarity(cat_vec, the_vec):.3f}")
print(f"  dog-the: {cosine_similarity(dog_vec, the_vec):.3f}")

# Load actual pre-trained GloVe vectors (commented out as it requires download)
"""
try:
    # Download from https://nlp.stanford.edu/projects/glove/
    glove_path = "glove.6B.100d.txt"
    glove_model = KeyedVectors.load_word2vec_format(glove_path, binary=False, no_header=False)
    
    print("\nUsing actual GloVe vectors:")
    print(f"Vector for 'king': {glove_model['king'][:10]}...")
    
    # Famous analogy
    result = glove_model.most_similar(positive=['woman', 'king'], negative=['man'], topn=3)
    print("\nking - man + woman = ?")
    for word, score in result:
        print(f"  {word}: {score:.3f}")
        
except FileNotFoundError:
    print("\nGloVe file not found. Please download from:")
    print("https://nlp.stanford.edu/projects/glove/")
"""
```

# Language Modeling (LM)

Language modeling or LM in NLP is a probabilistic and statistical methods used to determine the chances of a given series of words occurring in a text based on the previous words. It helps to predict which word is more likely to come next in the sentence.

Language modeling is useful for a variety of tasks like speech recognition, machine learning, handwritten recognition, natural language processing, natural language understanding and natural language generation (generating more human-like text). Particularly ones that perform text generation, machine translation and question answering.



Levels of Language Modeling
1. characters
2. word
3. ngram
4. "refrigerator magnets" model
5. phrases
6. sentences
7. conversation (groups of sentences)

### What is an n-gram?

An n-gram is a sequence of 'n' consecutive items (words, characters, or symbols) from a given text. It's like a sliding window that moves through the text to capture these sequences.

![N-grams in NLP](https://cdn.botpenguin.com/assets/website/N_Gram_feb7947286.png)

**Language Model Fundamentals**

 **Conditional Probability in Language Models**
Calculate $P(w|h)$, the probability that the next word is $w$, given context/history $h$.

**Example**
For the phrase: "This article is on...", if we want to predict the likelihood of "NLP" as the next word:

$P(\text{"NLP"} \mid \text{"This", "article", "is", "on"})$

**Chain Rule of Probability**
The probability of a sequence of words is computed as:

$$P(w_1, w_2, \dots, w_n) = \prod_{i=1}^{n} P(w_i \mid w_1, w_2, \dots, w_{i-1})$$

**Markov Assumption**
To reduce complexity, N-gram models assume the probability of a word depends only on the previous $n-1$ words:

$$P(w_i \mid w_1, \dots, w_{i-1}) \approx P(w_i \mid w_{i-(n-1)}, \dots, w_{i-1})$$

**Evaluating Language Models**

**1. Entropy**
Measures the uncertainty or information content in a distribution:

$$H(p) = \sum_x p(x) \cdot (-\log(p(x)))$$

- Always gives non-negative values

**2. Cross-Entropy**
Measures how well a probability distribution $q$ predicts a sample from test data with true distribution $p$:

$$H(p,q) = -\sum_x p(x) \log(q(x))$$

- Usually $\geq$ entropy
- Reflects model "surprise" at the test data

**3. Perplexity**
Exponential of cross-entropy; lower values indicate a better model:

$$\text{Perplexity}(W) = \sqrt[N]{\prod_{i=1}^{N} \frac{1}{P(w_i \mid w_{i-1})}}$$

Where:
- $N$ is the number of words in the test set
- $P(w_i \mid w_{i-1})$ is the bigram probability of word $w_i$ given $w_{i-1}$

```python
import nltk
from nltk import trigrams
from nltk.corpus import reuters
from collections import defaultdict

nltk.download('reuters')
nltk.download('punkt')

words = nltk.word_tokenize(' '.join(reuters.words()))
tri_grams = list(trigrams(words))

model = defaultdict(lambda: defaultdict(lambda: 0))
for w1, w2, w3 in tri_grams:
    model[(w1, w2)][w3] += 1

for w1_w2 in model:
    total_count = float(sum(model[w1_w2].values()))
    for w3 in model[w1_w2]:
        model[w1_w2][w3] /= total_count


def predict_next_word(w1, w2):
    next_word_probs = model[w1, w2]
    if next_word_probs:
        return max(next_word_probs, key=next_word_probs.get)
    else:
        return "No prediction available"


print("Next Word:", predict_next_word('the', 'stock'))
```

### Skip-Gram Model in Word2Vec

**Overview**
In the skip-gram architecture of word2vec, the model predicts context words based on a center word. Given an array of words W, if W(i) is the input (center word), then W(i-2), W(i-1), W(i+1), and W(i+2) are the context words when using a sliding window of size 2.

![Skip-gram Architecture](https://media.geeksforgeeks.org/wp-content/uploads/Skip-gram-architecture-2.jpg)

**Forward Propagation**

**1. Hidden Layer Calculation**
- Multiply one-hot encoding of the center word (x) with the first weight matrix W to get hidden layer matrix h:
  $$h = W^T \cdot x$$
- **Dimensions**: $(N \times V) \cdot (V \times 1) \rightarrow (N \times 1)$
  - V = Vocabulary size
  - N = Dimension of hidden layer

**2. Output Layer Calculation**
- Multiply hidden layer vector h with second weight matrix W' to get matrix u:
  $$u = W'^T \cdot h$$
- **Dimensions**: $(V \times N) \cdot (N \times 1) \rightarrow (V \times 1)$

**3. Softmax Activation**
- Apply softmax to get the output probabilities:
  $$u_j = V_{w_j}^T \cdot h$$
  $$y = \text{softmax}(u)$$
  $$y_j = \text{softmax}(u_j) = \frac{e^{u_j}}{\sum_{j'=1}^V e^{u_{j'}}}$$
  $$P(w_j | w_i) = y_j$$

**4. Objective**
Maximize the probability of context words given the center word:
$$\prod_{c=1}^C \frac{e^{u_{j_c^*}}}{\sum_{j'=1}^V e^{u_{j'}}}$$

**Loss Function**
Negative log-likelihood of the above probability:
$$E = -\log \left( \prod_{c=1}^C \frac{e^{u_{j_c^*}}}{\sum_{j'=1}^V e^{u_{j'}}} \right)$$
$$E = -\sum_{c=1}^C u_{j_c^*} + C \cdot \log \sum_{j'=1}^V e^{u_{j'}}$$

**Backpropagation**

**1. Derivatives for W' (Output Weights)**
$$\frac{\partial E}{\partial W'} = \frac{\partial E}{\partial u_j} \cdot \frac{\partial u_j}{\partial W'}$$

Where:
- $\frac{\partial E}{\partial u_j} = y_j - t_j = e_j$ (prediction error)
- $\frac{\partial E}{\partial W'} = e_j \cdot h_i$

**2. Derivatives for W (Input Weights)**
$$\frac{\partial E}{\partial W} = \frac{\partial E}{\partial u_j} \cdot \frac{\partial u_j}{\partial h_i} \cdot \frac{\partial h_i}{\partial W}$$

Which simplifies to:
$$\frac{\partial E}{\partial W} = e_j \cdot W' \cdot x_i$$

**Training**
- Update weights using gradient descent:
  $$W' = W' - \alpha \cdot \frac{\partial E}{\partial W'}$$
  $$W = W - \alpha \cdot \frac{\partial E}{\partial W}$$
  Where $\alpha$ is the learning rate.


```python
import numpy as np
import string
from nltk.corpus import stopwords 
import nltk
nltk.download('stopwords')
 
def softmax(x):
    """Compute softmax values for each sets of scores in x."""
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum()
 
class word2vec(object):
    def __init__(self):
        self.N = 10
        self.X_train = []
        self.y_train = []
        self.window_size = 2
        self.alpha = 0.001
        self.words = []
        self.word_index = {}
 
    def initialize(self,V,data):
        self.V = V
        self.W = np.random.uniform(-0.8, 0.8, (self.V, self.N))
        self.W1 = np.random.uniform(-0.8, 0.8, (self.N, self.V))
         
        self.words = data
        for i in range(len(data)):
            self.word_index[data[i]] = i
 
     
    def feed_forward(self,X):
        self.h = np.dot(self.W.T,X).reshape(self.N,1)
        self.u = np.dot(self.W1.T,self.h)
        #print(self.u)
        self.y = softmax(self.u)  
        return self.y
         
    def backpropagate(self,x,t):
        e = self.y - np.asarray(t).reshape(self.V,1)
        # e.shape is V x 1
        dLdW1 = np.dot(self.h,e.T)
        X = np.array(x).reshape(self.V,1)
        dLdW = np.dot(X, np.dot(self.W1,e).T)
        self.W1 = self.W1 - self.alpha*dLdW1
        self.W = self.W - self.alpha*dLdW
         
    def train(self,epochs):
        for x in range(1,epochs):        
            self.loss = 0
            for j in range(len(self.X_train)):
                self.feed_forward(self.X_train[j])
                self.backpropagate(self.X_train[j],self.y_train[j])
                C = 0
                for m in range(self.V):
                    if(self.y_train[j][m]):
                        self.loss += -1*self.u[m][0]
                        C += 1
                self.loss += C*np.log(np.sum(np.exp(self.u)))
            print("epoch ",x, " loss = ",self.loss)
            self.alpha *= 1/( (1+self.alpha*x) )
            
    def predict(self,word,number_of_predictions):
        if word in self.words:
            index = self.word_index[word]
            X = [0 for i in range(self.V)]
            X[index] = 1
            prediction = self.feed_forward(X)
            output = {}
            for i in range(self.V):
                output[prediction[i][0]] = i
             
            top_context_words = []
            for k in sorted(output,reverse=True):
                top_context_words.append(self.words[output[k]])
                if(len(top_context_words)>=number_of_predictions):
                    break
     
            return top_context_words
        else:
            print("Word not found in dictionary")
          


def preprocessing(corpus):
    stop_words = set(stopwords.words('english'))    
    training_data = []
    sentences = corpus.split(".")
    for i in range(len(sentences)):
        sentences[i] = sentences[i].strip()
        sentence = sentences[i].split()
        x = [word.strip(string.punctuation) for word in sentence
                                     if word not in stop_words]
        x = [word.lower() for word in x]
        training_data.append(x)
    return training_data
     
 
def prepare_data_for_training(sentences,w2v):
    data = {}
    for sentence in sentences:
        for word in sentence:
            if word not in data:
                data[word] = 1
            else:
                data[word] += 1
    V = len(data)
    data = sorted(list(data.keys()))
    vocab = {}
    for i in range(len(data)):
        vocab[data[i]] = i
     
    #for i in range(len(words)):
    for sentence in sentences:
        for i in range(len(sentence)):
            center_word = [0 for x in range(V)]
            center_word[vocab[sentence[i]]] = 1
            context = [0 for x in range(V)]
            
            for j in range(i-w2v.window_size,i+w2v.window_size):
                if i!=j and j>=0 and j<len(sentence):
                    context[vocab[sentence[j]]] += 1
                    w2v.X_train.append(center_word)
                    w2v.y_train.append(context)
    w2v.initialize(V,data)
 
    return w2v.X_train,w2v.y_train   


corpus = ""
corpus += "The earth revolves around the sun. The moon revolves around the earth"
epochs = 1000

training_data = preprocessing(corpus)
w2v = word2vec()

prepare_data_for_training(training_data,w2v)
w2v.train(epochs) 

print(w2v.predict("around",3))
```

## BERT Model 

BERT (Bidirectional Encoder Representations from Transformers) leverages a transformer-based neural network to understand and generate human-like language. BERT employs an encoder-only architecture. In the original Transformer architecture, there are both encoder and decoder modules. The decision to use an encoder-only architecture in BERT suggests a primary emphasis on understanding input sequences rather than generating output sequences.

Traditional language models process text sequentially, either from left to right or right to left. This method limits the model's awareness to the immediate context preceding the target word. BERT uses a bi-directional approach considering both the left and right context of words in a sentence, instead of analyzing the text sequentially, BERT looks at all the words in a sentence simultaneously.

### Text to text Transfer Transformer (T5)

Text-to-Text Transfer Transformer (T5) is a large transformer model trained on the Colossal Clean Crawled Corpus (C4). It was released as a pre-trained model capable of handling various NLP tasks such as translation, summarization, question answering and classification.

There are multiple ways to use the T5 (Text-to-Text Transfer Transformer) model for data augmentation in NLP tasks.

1. Using T5 Directly
Similar to back translation, T5 can be used without additional training by leveraging its pre-trained summarization capabilities. In this approach:

The input is given in the format: "summarize: <input text>"
T5 generates an abstractive summary, often rephrasing or using new words.
This is useful for long-text NLP tasks like document classification or summarization.
However, for short texts, the quality of augmented data may not be very effective.
2. Fine-Tuning T5 for Custom Data Augmentation
T5 can also be fine-tuned on specific tasks to generate high-quality synthetic data. Two effective strategies are:

1. Masked Span Prediction
T5 can be fine-tuned similarly to BERT for masked language modeling.
Input format: "predict mask: The [MASK] barked at the stranger."
Output: "The dog barked at the stranger."
You can mask multiple words (spans) to generate more diverse sentence structures.
This helps produce augmented text with structural variations, mimicking BERT-style augmentation.
![image](https://media.geeksforgeeks.org/wp-content/uploads/20210102163949/img2.jpg)

2. Paraphrase Generation
T5 can be fine-tuned to create paraphrases that retain meaning but vary in structure and wording.
The PAWS dataset is commonly used for this task.
Training involves formatting input as: "generate paraphrase: <sentence>" and output as its paraphrase.
The model can generate multiple variations, helping expand and diversify NLP datasets.
![image](https://media.geeksforgeeks.org/wp-content/uploads/20210102164040/img3-660x280.jpg)


```python
%pip install transformers torch datasets

import pandas as pd
import numpy as np
import torch
from transformers import T5Tokenizer, T5ForConditionalGeneration

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
# load Paraphrasing Model
class T5Paraphraser:
    def __init__(self, model_name="ramsrigouthamg/t5_paraphraser"):
        self.tokenizer = T5Tokenizer.from_pretrained(model_name)
        self.model = T5ForConditionalGeneration.from_pretrained(model_name).to(device)

    def paraphrase(self, text, num_return_sequences=3, max_length=128):
        input_text = f"paraphrase: {text} </s>"
        inputs = self.tokenizer.encode(input_text, return_tensors="pt", max_length=max_length, truncation=True).to(device)

        outputs = self.model.generate(
            inputs, max_length=max_length, num_return_sequences=num_return_sequences,
            do_sample=True, top_k=50, top_p=0.95, temperature=0.8
        )

        return list(set(self.tokenizer.decode(output, skip_special_tokens=True) for output in outputs))
#Initialise model
paraphraser = T5Paraphraser()

sample_sentences = [
    "Python is a powerful language.",
    "Deep learning requires large datasets.",
    "Artificial intelligence is evolving rapidly."
]

for text in sample_sentences:
    print(f"\nOriginal: {text}")
    for i, p in enumerate(paraphraser.paraphrase(text), 1):
        print(f"Paraphrase {i}: {p}")

#Augmented a Text Classification Dataset
def get_sample_data():
    return pd.DataFrame({
        'text': [
            "Great product quality", "Fast delivery", "Excellent service",
            "Poor design", "Terrible support"
        ],
        'label': ['positive', 'positive', 'positive', 'negative', 'negative']
    })

def augment_dataset(df, text_col='text', label_col='label', ratio=0.5):
    augmented = []

    for label in df[label_col].unique():
        samples = df[df[label_col] == label].sample(frac=ratio)
        for text in samples[text_col]:
            for p in paraphraser.paraphrase(text, num_return_sequences=2):
                augmented.append({text_col: p, label_col: label, 'source': 'augmented'})
    
    df['source'] = 'original'
    return pd.concat([df, pd.DataFrame(augmented)], ignore_index=True)

df = get_sample_data()
aug_df = augment_dataset(df)

print("\nAugmented Dataset:")
print(aug_df.head())
```

### Word Embedding using Word2Vec

Word2Vec is a word embedding technique in natural language processing (NLP) that allows words to be represented as vectors in a continuous vector space. Researchers at Google developed word2Vec that maps words to high-dimensional vectors to capture the semantic relationships between words. It works on the principle that words with similar meanings should have similar vector representations. Word2Vec utilizes two architectures:

1. CBOW (Continuous Bag of Words)
The CBOW model predicts the current word given context words within a specific window. The input layer contains the context words and the output layer contains the current word. The hidden layer contains the dimensions we want to represent the current word present at the output layer. 

![CBOW](https://media.geeksforgeeks.org/wp-content/uploads/20250812155937478236/CBOW.webp)

2. Skip Gram
The Skip gram predicts the surrounding context words within specific window given current word. The input layer contains the current word and the output layer contains the context words. The hidden layer contains the number of dimensions in which we want to represent current word present at the input layer.
![Skip Gram](https://media.geeksforgeeks.org/wp-content/uploads/20250813163102869325/output.webp)
