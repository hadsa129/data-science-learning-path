```python
# collections module
import collections

#Counter: is a dict subclass for counting hashable objects.


#collections.Counter(iterable:list):it counts each element in the iterable.
collections.Counter([1, 2, 2, 3, 3, 3])
#collections.Counter(mapping:dict) :Keys are the items, and values are the counts.
collections.Counter({"a": 4, "b": 2})
#collections.Counter(**kwds):Using keyword arguments
c = collections.Counter(a=4, b=2)
#to count the number of times each element appears in a list
collections.Counter([1, 2, 2, 3, 3, 3])

#to get the most common element :Return a list of the n most common elements and their counts
collections.Counter([1, 2, 2, 3, 3, 3]).most_common(5)

# to get the elements
collections.Counter([1, 2, 2, 3, 3, 3]).elements()

#to subtract counts:subtracted from an iterable or from another Counter
c = collections.Counter(a=4, b=2, c=0, d=-2)
d = collections.Counter(a=1, b=2, c=3, d=4)
c.subtract(d)

#to get the total count of all elements
c.total()

#to clear the counter: remove all elements
c.clear()

#defaultdict : dict with a default value for missing keys
#collections.defaultdict(default_factory:callable):default_factory is a function that provides the default value for missing keys.
collections.defaultdict(int)  # default value is 0
collections.defaultdict(list)  # default value is empty list
collections.defaultdict(set)  # default value is empty set


```

```python
#re module :regular expressions:
import re
#regex:
#r"\d+":one or more digits
#r"\w+":one or more word characters
#r"\s+":one or more whitespace characters
#\D:not digits exp:\D → "!", " "
#\W:not word characters exp:\W → "!", " "
#\S:not whitespace characters exp \S → "a", "5", "!"
#*:occurs zero or more times exp: \d* → "", "123", "7"
#+:occurs one or more times exp: \d+ → "123", "7"
#?: 0 or 1 occurrence exp: (colou?r) = color or colour
#^:start of string exp: ^abc → "abc", "abc123"
#$:end of string exp: abc$ → "abc", "123abc"
#{m}:occurs m times exp: \d{3} → "123"
#{m,n}:occurs between m and n times exp: \d{2,4} → "12", "123", "1234"
#|:or exp: x|y → Matches either "x" or "y"
#x(?=y):Matches "x" only if it is followed by "y"
#x(?!y):Matches "x" only if it is not followed by "y"
#(?<=y)x :Matches "x" only if "x" is preceded by "y".
#\b:word boundary exp: \babc → "abc", "abc123"
#to search for a pattern in a string:
re.search(r"\d+", "I have 2 apples and 10 oranges.") #returns a match object if the pattern is found, otherwise None

#to find all occurrences of a pattern in a string:
re.findall(r"\d+", "I have 2 apples and 10 oranges.") #returns a list of all matches

#to match a pattern at the beginning of a string:
re.match(r"have", "I have 2 apples and 10 oranges.") #returns a match object if the pattern is found at the beginning of the string, otherwise None

#to substitute a pattern in a string:
re.sub(r"\d+", "X", "I have 2 apples and 10 oranges.") #returns a new string with all occurrences of the pattern replaced with the replacement =>'I have X apples and X oranges.'

#to split a string:
re.split(r"\d+", "I have 2 apples and 10 oranges.") #returns a list of substrings split by the pattern :['I have ', ' apples and ', ' oranges.']
```

# Functions

```python
#Lambda functions :anonymous function that can have any number of arguments but only one expression
#Syntax -> lambda argument : expression
sum_of_two_number = lambda x,y : x+y
sum_of_two_number(1,2)
pow_x = lambda x : x**2
pow_x(2)
```

```python
# Map function: Map applies the supplied function to all items in the iterable(s) passed to it. It returns a new iterable of the results generated by the applied function.
#map(function, iterable, *iterables)

#map with one iterable
nombres = [1, 2, 3, 4, 5]
carres = map(lambda x: x**2, nombres)
print(list(carres))  # [1, 4, 9, 16, 25]

#map with multiple iterables
nombres = [1, 2, 3, 4, 5]
lettres = ['a', 'b', 'c', 'd', 'e']
resultat = map(lambda x, y: (x, y), nombres, lettres)
print(list(resultat))  # [(1, 'a'), (2, 'b'), (3, 'c'), (4, 'd'), (5, 'e')]

```

```python
#Filter function: Filter applies a function to all items in an iterable and returns a new iterable containing only the items for which the function returns True.
#filter(function, iterable)

#Filter with function
def is_even(x):
    return x%2==0
b = [2,5,4,22,13,122,34,25,21]
lst = list(filter(is_even, b))

#Filter with lambda function
lst = list(filter(lambda x: x%2==0, b))
```

```python
#Reduce function: Reduce applies a function to all items in an iterable and returns a single value.
#reduce(function, iterable)

from functools import reduce
def add(x,y):
    return x+y
c = [1,2,3,4,5,6]
total = reduce(add, c)
print(total)
```
