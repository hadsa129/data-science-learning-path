```python
#apply a function
def set_age_category(age):
    if age < 30:
        return 'jeune'
    elif age < 50:
        return 'moyen'
    else:
        return 'vieux'

employees_df_copy['âge_categorie'] = employees_df_copy['âge'].apply(set_age_category)
#using pd.cut:cut the data into bins
employees_df['âge_categorie'] = pd.cut(employees_df['âge'], bins=[0, 30, 50, 100], labels=['jeune', 'moyen', 'vieux'])
# using pd.qcut:cut thedata into quantiles
employees_df['âge_categorie'] = pd.qcut(employees_df['âge'], q=3, labels=['jeune', 'moyen', 'vieux'])
```

```python
#aggregation multiple colonnes
agg_gen=employees_df.groupby('département').agg(
    {'salaire':lambda x: pd.to_numeric(x, errors='coerce').mean() ,
    'âge':lambda x: pd.to_numeric(x, errors='coerce').mean(),
    'expérience':lambda x: pd.to_numeric(x, errors='coerce').mean()
    }
).rename(columns={'salaire':'moy_salaire', 'âge':'moy_âge', 'expérience':'moy_expérience'})
```

```python
#merge dataframes:merging is like to join tables in sql
clients_data = {
    'client_id': [1, 2, 3, 4, 5],
    'nom': ['Alice Dupont', 'Bob Martin', 'Charlie Brown', 'Diana Prince', 'Eve Smith'],
    'ville': ['Paris', 'Lyon', 'Marseille', 'Paris', 'Lyon'],
    'date_inscription': ['2023-01-15', '2023-02-20', '2023-01-10', '2023-03-05', '2023-02-28']
}

transactions_data = {
    'transaction_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'client_id': [1, 1, 2, 3, 1, 4, 2, 3, 5, 1],
    'montant': [150, 200, 75, 300, 50, 120, 180, 90, 250, 100],
    'date_transaction': ['2023-02-01', '2023-03-15', '2023-02-28', '2023-03-10', 
                        '2023-04-01', '2023-03-20', '2023-04-15', '2023-05-01',
                        '2023-04-10', '2023-05-20']
}
clients_df = pd.DataFrame(clients_data)
transactions_df = pd.DataFrame(transactions_data)

#inner join :to keep only the rows that have a match in both dataframes
inner_join = pd.merge(clients_df, transactions_df, on='client_id', how='inner')
print(inner_join)

#left join :to keep all the rows from the left dataframe and the matching rows from the right dataframe
left_join = pd.merge(clients_df, transactions_df, on='client_id', how='left')
print(left_join)

#right join :to keep all the rows from the right dataframe and the matching rows from the left dataframe
right_join = pd.merge(clients_df, transactions_df, on='client_id', how='right')
print(right_join)

#outer join :to keep all the rows from both dataframes
outer_join = pd.merge(clients_df, transactions_df, on='client_id', how='outer')
print(outer_join)



```
